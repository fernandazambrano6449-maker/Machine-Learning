## Medidas descriptivas




```python
#Calcular media

df_titanic["age"].mean()
df_titanic["fare"].mean()

print(df_titanic["age"].mean())
print(df_titanic["fare"].mean())
```

```python
# Mediana
df_titanic["age"].median()
df_titanic["fare"].median()

print(df_titanic["age"].median())
print(df_titanic["fare"].median())
```

```python
# Moda
df_titanic["age"].mode()
df_titanic["fare"].mode()

print(df_titanic["age"].mode())
print(df_titanic["fare"].mode())
```

```python
# Desviación estándar
df_titanic["age"].std()
df_titanic["fare"].std()

print(df_titanic["age"].std())
print(df_titanic["fare"].std())
```

```python
# Cuantiles
df_titanic["age"].quantile([0.25, 0.5, 0.75])
df_titanic["fare"].quantile([0.25, 0.5, 0.75])

print(df_titanic["age"].quantile([0.25, 0.5, 0.75]))
print(df_titanic["fare"].quantile([0.25, 0.5, 0.75]))
```

Interpretación de edad: la media (29.88) y la mediana (28) están muy cerca por que hay una distribución bastante simétrica, sin muchos valores extremos jalando el promedio.

Interpretación de tarifa: la media (33.30) es más del doble que la mediana (14.45). Eso es una señal de asimetría hacia la derecha, hay algunos pasajeros que pagaron tarifas muy altas que aumentan el promedio, mientras que la mayoría de la gente pagó bastante menos. La desviación estándar de fare confirma que hay mucha dispersión.

## Tablas de frecuencia y cruces de variables


```python
# Distribución de pasajeros por clase
df_titanic["pclass"].value_counts()
```

Hubo más pasajero de tercera clase

```python
# Tabla cruzada entre sexo y supervivencia
pd.crosstab(df_titanic["sex"], df_titanic["survived"])
```

El sexo infliyó mucho en la tasa de superviviencia. El 72% sobrevivió mientras que solo el 19% de los hombres sobrevivieron  

```python
# Tabla cruzada entre clase y supervivencia
pd.crosstab(df_titanic["pclass"], df_titanic["survived"])
```

A menor clase, menor tasa de supervivencia. Esto es consistente con factores históricos documentado ya que los camarotes de 1ra clase estaban más cerca de los botes salvavidas, y hubo diferencias en el acceso a la evacuación según la clase socioeconómica.

## Visualizaciones



```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Histograma de Age
plt.hist(df_titanic["age"].dropna())
plt.title("Distribución de la edad")
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.show()
```

Podemos ver que la mayoría de personas tenían entre los 20 y 30 años, mientras que muy pocas personas estaban entre los 70 y 80 años

```python
# Histograma de Fare
plt.hist(df_titanic["fare"])
plt.title("Distribución de la tarifa")
plt.xlabel("Fare")
plt.ylabel("Frecuencia")
plt.show()
```

Vemos que hay una hay más de 1,000 pasajeros que pagaron un boleto entre 0 y 50 dólares. Después las barras disminuyes significativamente hacia la derecha ya que menos de 50 personas pagaron entre 100 y 150 y unas 4 o 5 personas pagaron más de 500 dólares

```python
sns.boxplot(x="pclass", y="fare", data=df_titanic)
plt.title("Tarifa por clase")
plt.show()
```

El boxplot muestra que la clase 1 no solo tuvo tarifas más altas en general, sino también mucha más variabilidad de precios. Las clases 2 y 3 tienen cajas pequeñas y comprimidas cerca de valores bajos, lo que indica que la mayoría de esos pasajeros pagó tarifas similares entre sí, con poca dispersión

```python
sns.countplot(x="sex", hue="survived", data=df_titanic)
plt.title("Supervivencia según sexo")
plt.show()
```

```python
corr = df_titanic.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de calor de correlación")
plt.show()
```

Existe una correlación débil-moderada negativa entre pclass y survived (r = -0.31), sugiriendo que a peor clase, menor probabilidad de supervivencia
